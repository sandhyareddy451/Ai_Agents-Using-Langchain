```python
import langchain
langchain.__version__
```

```python
from langchain.agents import initialize_agent,Tool
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
GROQ_API_KEY = os.getenv('GROQ_API_KEY')
os.environ['GROQ_API_KEY'] = GROQ_API_KEY
```

```python
from langchain_community.tools.ddg_search.tool import DuckDuckGoSearchRun
search_tool = Tool(
    
 name = "DuckDuckGo Search",
 func = DuckDuckGoSearchRun().run,
 description="Use thos tool to perform web searches using DuckDuckGo"
)
llm = ChatGroq(temperature=0.5,  model="llama3-70b-8192",)
tool = [search_tool]
agent = initialize_agent(tool,llm,agent = "zero-shot-react-description", verbose =True)
response = agent.run("what are the top 5 rated movies in 2023")
```

### SQLite database

```python
import sqlite3
connection = sqlite3.connect("example.db")
cursor = connection.cursor()
cursor.execute(""" CREATE TABLE IF NOT EXISTS inventory(
               id  INTEGER PRIMARY KEY,
               product_name TEXT NOT NULL,
               stock_quality INTEGER NOT NULL
)             
""")

cursor.executemany("""INSERT INTO inventory (product_name , stock_quality) VALUES (?,?)
                   """, [
                       ("Product A", 100),
                       ("Product B", 500),
                       ("Product C", 200),
                       ("Product D", 300),
                       ("Product E", 150),
                       
])
connection.commit()
connection.commit()
```
